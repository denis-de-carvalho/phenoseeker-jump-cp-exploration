# Locating image files on S3 and understanding the path structure

The full JUMP-CP image data is hosted in a public bucket: `s3://cellpainting-gallery/cpg0016-jump/`. The goal of this notebook is to find where the images of my three selected plates live in that bucket, and to understand how the hierarchy (source, batch, plate, well, imaging site, fluorescence channel) is reflected in the file paths.

No image is downloaded here; this notebook only manipulates metadata and S3 file listings.

```python
# Install s3fs so pandas can read S3 directly (anonymous, public bucket)
!pip install s3fs --quiet

import pandas as pd
import s3fs

# Reuse the working subset selected in notebook 02
SUBSET_SOURCE = "source_3"
SUBSET_BATCH  = "CP59"
SUBSET_PLATES = ["BR5867a3", "BR5867b3", "BR5867c3"]

# Anonymous filesystem handle for the public S3 bucket
fs = s3fs.S3FileSystem(anon=True)
BUCKET = "cellpainting-gallery"

print(f"Subset: {SUBSET_SOURCE}, batch {SUBSET_BATCH}, plates {SUBSET_PLATES}")
```

```python
fs.ls(f"{BUCKET}/cpg0016-jump/{SUBSET_SOURCE}/workspace/load_data_csv/{SUBSET_BATCH}/")[:20]
```

```python
plate_dir = f"{BUCKET}/cpg0016-jump/{SUBSET_SOURCE}/workspace/load_data_csv/{SUBSET_BATCH}/{SUBSET_PLATES[0]}/"
fs.ls(plate_dir)
```

```python
# Take whichever load_data file is in the directory
load_data_files = [f for f in fs.ls(plate_dir) if "load_data" in f and f.endswith(".csv")]
print(f"Files found: {load_data_files}")

# Pick the simplest one (without illumination correction info)
load_data_path = "s3://" + [f for f in load_data_files if "illum" not in f][0]
print(f"Loading: {load_data_path}")

load_data = pd.read_csv(load_data_path, storage_options={"anon": True})
print(f"Shape: {load_data.shape}")
print(f"Columns: {list(load_data.columns)}")
load_data.head()
```

```python
# How many imaging sites per well?
sites_per_well = load_data.groupby("Metadata_Well")["Metadata_Site"].nunique()
print(f"Number of wells in this plate: {len(sites_per_well)}")
print(f"Sites per well (typical values):")
print(sites_per_well.value_counts())
```

```python
# Inspect what a single row contains: one well, one site, multiple channels
single_row = load_data.iloc[0]
print("Identifying columns:")
print(single_row[["Metadata_Plate", "Metadata_Well", "Metadata_Site"]])
print()
print("Image file paths and names:")
for col in load_data.columns:
    if col.startswith("PathName_Orig") or col.startswith("FileName_Orig"):
        print(f"  {col}: {single_row[col]}")
```

```python
# What image-related columns does this load_data file actually have?
image_cols = [c for c in load_data.columns if any(p in c for p in ["Path", "File", "URL", "Channel"])]
print("Image-related columns found in this load_data:")
for c in image_cols:
    print(f"  {c}")
```

```python
# For one specific well/site, show the S3 path to each of the 5 fluorescence channels

url_cols = sorted([c for c in load_data.columns if c.startswith("URL_Orig")])

print(f"Identifying info:")
print(f"  Plate: {single_row['Metadata_Plate']}")
print(f"  Well:  {single_row['Metadata_Well']}")
print(f"  Site:  {single_row['Metadata_Site']}")
print()
print("Full S3 paths to each fluorescence channel at this position:")
for col in url_cols:
    channel = col.replace("URL_Orig", "")
    print(f"  {channel:5s} -> {single_row[col]}")
```

## Summary

### The structure of JUMP-CP on S3

The JUMP-CP data is in the public AWS bucket `s3://cellpainting-gallery/cpg0016-jump/`. At the top, I find one folder per data source: `source_1` through `source_15`, with `source_12` and `source_14` missing as expected from the metadata, plus a `source_all` folder that I have not investigated yet.

Inside each source, three folders:

- `images/`: the actual TIFF files
- `workspace/`: CellProfiler outputs, metadata tables, and the "load_data" index files that I need
- `workspace_dl/`: deep learning related outputs, most likely pre-computed feature embeddings (worth coming back to later)

### Where image paths are listed

The index files live in `workspace/load_data_csv/<batch>/<plate>/`. For my batch `CP59` I find 20+ plates including my 3 selected ones (`BR5867a3`, `BR5867b3`, `BR5867c3`). Each plate folder contains two files: `load_data.csv` (the basic index) and `load_data_with_illum.csv` (same index plus illumination correction file paths, which I do not need yet).

### What's inside a load_data.csv

For plate `BR5867a3`, the `load_data.csv` has **3,456 rows and 34 columns**. The 3,456 rows correspond exactly to **384 wells × 9 imaging sites per well**. Every well of this plate is imaged at 9 different positions by the microscope, so I will need to handle this 9-fold replication every time I work on a single well.

Each row represents one (well, site) combination, with these key columns:

- `Metadata_Plate`, `Metadata_Well`, `Metadata_Site`: identifying the row
- 5 columns `URL_OrigXXX` (one per Cell Painting fluorescence channel): each one gives the full S3 path to one image file

The 5 channels of the Cell Painting assay are: **DNA** (nuclei), **RNA**, **ER** (endoplasmic reticulum), **AGP** (actin / golgi / plasma membrane), **Mito** (mitochondria). One (well, site) combination produces 5 image files, one per channel.

A note on column naming convention: in `source_3`, each image is referenced by a single `URL_OrigXXX` column that contains the full S3 path. Other sources of JUMP-CP can use a `PathName_*` + `FileName_*` pair instead. The column convention is therefore worth checking when changing source.

### How the image files are named

A full S3 path looks like this:

s3://cellpainting-gallery/cpg0016-jump/source_3/images/CP59/images/BR5867a3__2022-04-29T01_02_16-Measurement 1/Images/r01c01f01p01-ch5sk1fk1fl1.tiff

Breaking this down:

- The plate folder name includes the acquisition timestamp: `BR5867a3__2022-04-29T01_02_16-Measurement 1`. So plate BR5867a3 was imaged on 29 April 2022.
- The filename `r01c01f01p01-ch5sk1fk1fl1.tiff` encodes the position compactly:
  - `r01c01` = row 1, column 1 of the plate, i.e. well A01
  - `f01` = imaging field (site) number 1
  - `p01` = z-plane 1 (Cell Painting is 2D imaging, so always 1)
  - `ch5` = channel number 5

An important subtlety: the channel number in the filename does NOT correspond directly to the channel name listed in the CSV column. The mapping I observed for this plate is:

| Filename code | Channel name |
|:-------------:|:------------:|
| ch1           | Mito         |
| ch2           | AGP          |
| ch3           | RNA          |
| ch4           | ER           |
| ch5           | DNA          |

So it is safer to always use the `URL_OrigXXX` column from the load_data CSV than to try to parse the filename. The CSV column already tells me which channel is which.
